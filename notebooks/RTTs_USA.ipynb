### Two-window algorithm

```python
def count_flows_with_surge(rtt_dict, win_size, surge_thresh, agg=False):

    # Keep track of flow IDs and their detection status
    detection_dict = {}
    
    # Keep track of flow counts with surges
    qualified_count = 0
    surged_count = 0
    
    for flow_id in rtt_dict:
        
        # Extract RTTs for the flow
        flow_rtts = rtt_dict[flow_id]
        # Get mapped IP prefix
        if not agg:
            prefix = flowid_prefix_map[flow_id]
        else:
            prefix = flow_id
        # Set detection status to False
        flagged = False
        
        # Don't process the flow if there aren't enough samples
        if len(flow_rtts) < 2 * win_size:
            continue
        qualified_count += 1
        
        # Compute minrtt from first window
        minrtt_w1 = min(flow_rtts[:win_size])

        # Check for surges in subsequent windows
        for i in range(win_size, len(flow_rtts), win_size):
            minrtt_w2 = min(flow_rtts[i:i+win_size])
            surge = abs(minrtt_w1 - minrtt_w2)
            if surge >= surge_thresh:
                surged_count += 1
                flagged = True
                break
            minrtt_w1 = minrtt_w2 - 0.0

        if prefix not in detection_dict:
            detection_dict[prefix] = flagged
        elif not flagged and detection_dict[prefix]:
            detection_dict[prefix] = False

    return qualified_count, surged_count, len(detection_dict), sum(1 for fid in detection_dict if detection_dict[fid])
```

### Three window algorithm

```python
def count_flows_with_surge_and_stability(rtt_dict, win_size, surge_thresh, stable_thresh, agg=False):

    # Keep track of flow IDs and their detection status
    detection_dict = {}
    
    # Keep track of flow counts with surges
    qualified_count = 0
    flagged_count = 0
    
    for flow_id in rtt_dict:
        
        # Extract RTTs for the flow
        flow_rtts = rtt_dict[flow_id]
        # Get mapped IP prefix
        if not agg:
            prefix = flowid_prefix_map[flow_id]
        else:
            prefix = flow_id
        # Set detection status to False
        flagged = False
        
        # Don't process the flow if there aren't enough samples
        if len(flow_rtts) < 3 * win_size:
            continue
        qualified_count += 1

        # Compute minrtt from first window
        minrtt_w1 = min(flow_rtts[:win_size])
        # Compute minrtt from second window
        minrtt_w2 = min(flow_rtts[win_size:2*win_size])

        # Check for surges and stability in subsequent windows
        for i in range(2*win_size, len(flow_rtts), win_size):
            # Compute minrtt from third window
            minrtt_w3 = min(flow_rtts[i:i+win_size])
            # Compute surge
            surge = abs(minrtt_w1 - minrtt_w2)
            # Compute stability
            stability = abs(minrtt_w2 - minrtt_w3)
            # Check for surge and stability
            if surge >= surge_thresh:
                if stability <= stable_thresh:
                    flagged_count += 1
                    flagged = True
                    break
            minrtt_w1 = minrtt_w2 - 0.0
            minrtt_w2 = minrtt_w3 - 0.0

        if prefix not in detection_dict:
            detection_dict[prefix] = flagged
        elif not flagged and detection_dict[prefix]:
            detection_dict[prefix] = False

    return qualified_count, flagged_count, len(detection_dict), sum(1 for fid in detection_dict if detection_dict[fid])
```

```python
def extract_3windows_min_fp_for_variable(var):

    if var != "window" and var != "surge" and var != "stability":
        print("var provided is not supported!")
        return None
    
    x_3windows_window_size = []
    x_3windows_surge_threshold = []
    x_3windows_stability_threshold = []
    # Declare for all
    y_3windows_false_positive_count_all = []
    y_3windows_qualifying_flows_count_all = []
    y_3windows_prefixagg_false_positive_count_all = []
    y_3windows_prefixagg_qualifying_prefixes_count_all = []
    # Declare for PU clients
    y_3windows_false_positive_count_pucli = []
    y_3windows_qualifying_flows_count_pucli = []
    y_3windows_prefixagg_false_positive_count_pucli = []
    y_3windows_prefixagg_qualifying_prefixes_count_pucli = []
    # Declare for PU servers
    y_3windows_false_positive_count_pusrv = []
    y_3windows_qualifying_flows_count_pusrv = []
    y_3windows_prefixagg_false_positive_count_pusrv = []
    y_3windows_prefixagg_qualifying_prefixes_count_pusrv = []

    if var == "window":
        for wsz in [1] + list(range(STEP_WINDOW_SIZE, MAX_WINDOW_SIZE+1, STEP_WINDOW_SIZE)):
            x_3windows_window_size.append(wsz)
            # Initialize for all
            y_3windows_false_positive_count_all.append(10**8)
            y_3windows_qualifying_flows_count_all.append(10**8)
            y_3windows_prefixagg_false_positive_count_all.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_all.append(10**8)
            # Initialize for PU clients
            y_3windows_false_positive_count_pucli.append(10**8)
            y_3windows_qualifying_flows_count_pucli.append(10**8)
            y_3windows_prefixagg_false_positive_count_pucli.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_pucli.append(10**8)
            # Initialize for PU servers
            y_3windows_false_positive_count_pusrv.append(10**8)
            y_3windows_qualifying_flows_count_pusrv.append(10**8)
            y_3windows_prefixagg_false_positive_count_pusrv.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_pusrv.append(10**8)
            
    elif var == "surge":
        for srth in range(STEP_SURGE_THRESHOLD, MAX_SURGE_THRESHOLD+1, STEP_SURGE_THRESHOLD):
            x_3windows_surge_threshold.append(srth)
            # Initialize for all
            y_3windows_false_positive_count_all.append(10**8)
            y_3windows_qualifying_flows_count_all.append(10**8)
            y_3windows_prefixagg_false_positive_count_all.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_all.append(10**8)
            # Initialize for PU clients
            y_3windows_false_positive_count_pucli.append(10**8)
            y_3windows_qualifying_flows_count_pucli.append(10**8)
            y_3windows_prefixagg_false_positive_count_pucli.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_pucli.append(10**8)
            # Initialize for PU servers
            y_3windows_false_positive_count_pusrv.append(10**8)
            y_3windows_qualifying_flows_count_pusrv.append(10**8)
            y_3windows_prefixagg_false_positive_count_pusrv.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_pusrv.append(10**8)

    elif var == "stability":
        for stth in range(STEP_STABILITY_THRESHOLD, MAX_STABILITY_THRESHOLD+1, STEP_STABILITY_THRESHOLD):
            x_3windows_stability_threshold.append(stth)
            # Initialize for all
            y_3windows_false_positive_count_all.append(10**8)
            y_3windows_qualifying_flows_count_all.append(10**8)
            y_3windows_prefixagg_false_positive_count_all.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_all.append(10**8)
            # Initialize for PU clients
            y_3windows_false_positive_count_pucli.append(10**8)
            y_3windows_qualifying_flows_count_pucli.append(10**8)
            y_3windows_prefixagg_false_positive_count_pucli.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_pucli.append(10**8)
            # Initialize for PU servers
            y_3windows_false_positive_count_pusrv.append(10**8)
            y_3windows_qualifying_flows_count_pusrv.append(10**8)
            y_3windows_prefixagg_false_positive_count_pusrv.append(10**8)
            y_3windows_prefixagg_qualifying_prefixes_count_pusrv.append(10**8)

    for wsz in [1] + list(range(STEP_WINDOW_SIZE, MAX_WINDOW_SIZE+1, STEP_WINDOW_SIZE)):
        for srth in range(STEP_SURGE_THRESHOLD, MAX_SURGE_THRESHOLD+1, STEP_SURGE_THRESHOLD):
            for stth in range(STEP_STABILITY_THRESHOLD, MAX_STABILITY_THRESHOLD+1, STEP_STABILITY_THRESHOLD):
                if var == "window":
                    pos = x_3windows_window_size.index(wsz)
                elif var == "surge":
                    pos = x_3windows_surge_threshold.index(srth)
                elif var == "stability":
                    pos = x_3windows_stability_threshold.index(stth)
                    
                # Extract for all
                fp_count_all = results_3windows[(wsz, srth, stth)][1]
                qf_count_all = results_3windows[(wsz, srth, stth)][0]
                fp_prefixagg_count_all = results_3windows[(wsz, srth, stth)][3]
                qf_prefixagg_count_all = results_3windows[(wsz, srth, stth)][2]
                # Extract for PU clients
                fp_count_pucli = results_3windows_pucli[(wsz, srth, stth)][1]
                qf_count_pucli = results_3windows_pucli[(wsz, srth, stth)][0]
                fp_prefixagg_count_pucli = results_3windows_pucli[(wsz, srth, stth)][3]
                qf_prefixagg_count_pucli = results_3windows_pucli[(wsz, srth, stth)][2]
                # Extract for PU servers
                fp_count_pusrv = results_3windows_pusrv[(wsz, srth, stth)][1]
                qf_count_pusrv = results_3windows_pusrv[(wsz, srth, stth)][0]
                fp_prefixagg_count_pusrv = results_3windows_pusrv[(wsz, srth, stth)][3]
                qf_prefixagg_count_pusrv = results_3windows_pusrv[(wsz, srth, stth)][2]
                
                # Adjust for all
                if fp_count_all < y_3windows_false_positive_count_all[pos]:
                    y_3windows_false_positive_count_all[pos] = fp_count_all
                    y_3windows_qualifying_flows_count_all[pos] = qf_count_all
                if fp_prefixagg_count_all < y_3windows_prefixagg_false_positive_count_all[pos]:
                    y_3windows_prefixagg_false_positive_count_all[pos] = fp_prefixagg_count_all
                    y_3windows_prefixagg_qualifying_prefixes_count_all[pos] = qf_prefixagg_count_all
                # Adjust for PU clients
                if fp_count_pucli < y_3windows_false_positive_count_pucli[pos]:
                    y_3windows_false_positive_count_pucli[pos] = fp_count_pucli
                    y_3windows_qualifying_flows_count_pucli[pos] = qf_count_pucli
                if fp_prefixagg_count_pucli < y_3windows_prefixagg_false_positive_count_pucli[pos]:
                    y_3windows_prefixagg_false_positive_count_pucli[pos] = fp_prefixagg_count_pucli
                    y_3windows_prefixagg_qualifying_prefixes_count_pucli[pos] = qf_prefixagg_count_pucli
                # Adjust for PU servers
                if fp_count_pusrv < y_3windows_false_positive_count_pusrv[pos]:
                    y_3windows_false_positive_count_pusrv[pos] = fp_count_pusrv
                    y_3windows_qualifying_flows_count_pusrv[pos] = qf_count_pusrv
                if fp_prefixagg_count_pusrv < y_3windows_prefixagg_false_positive_count_pusrv[pos]:
                    y_3windows_prefixagg_false_positive_count_pusrv[pos] = fp_prefixagg_count_pusrv
                    y_3windows_prefixagg_qualifying_prefixes_count_pusrv[pos] = qf_prefixagg_count_pusrv

    if var == "window":
        var_values = x_3windows_window_size
    elif var == "surge":
        var_values = x_3windows_surge_threshold
    elif var == "stability":
        var_values = x_3windows_stability_threshold

    return var_values, \
        y_3windows_false_positive_count_all, y_3windows_qualifying_flows_count_all, \
        y_3windows_false_positive_count_pucli, y_3windows_qualifying_flows_count_pucli, \
        y_3windows_false_positive_count_pusrv, y_3windows_qualifying_flows_count_pusrv, \
        y_3windows_prefixagg_false_positive_count_all, y_3windows_prefixagg_qualifying_prefixes_count_all, \
        y_3windows_prefixagg_false_positive_count_pucli, y_3windows_prefixagg_qualifying_prefixes_count_pucli, \
        y_3windows_prefixagg_false_positive_count_pusrv, y_3windows_prefixagg_qualifying_prefixes_count_pusrv
```

### Determine the spread of max RTT surge and max RTT dip for flows to US servers and clients

```python
rtts_usa_pusrv_dict = df_rtts_usa_pusrv.groupby('Flow_ID')['RTT_ms'].apply(list).to_dict()
rtts_usa_pucli_dict = df_rtts_usa_pucli.groupby('Flow_ID')['RTT_ms'].apply(list).to_dict()
```

```python
def compute_rtt_maxsurge_maxdip(rtt_dict):
    maxsurge_perflow = {}
    maxdip_perflow = {}
    mismatch_count = 0
    for flow_id in rtt_dict:
        first_rtt = rtt_dict[flow_id][0]
        for second_rtt in rtt_dict[flow_id][1:]:
            surge = second_rtt - first_rtt
            dip = first_rtt - second_rtt
            if flow_id not in maxsurge_perflow:
                maxsurge_perflow[flow_id] = surge
            elif surge > maxsurge_perflow[flow_id]:
                maxsurge_perflow[flow_id] = surge
            if flow_id not in maxdip_perflow:
                maxdip_perflow[flow_id] = dip
            elif dip > maxdip_perflow[flow_id]:
                maxdip_perflow[flow_id] = dip
            first_rtt = second_rtt + 0.0
        if maxsurge_perflow[flow_id] - maxdip_perflow[flow_id] > 1.0:
            mismatch_count += 1
    return maxsurge_perflow, maxdip_perflow, mismatch_count
```

### Determine spread of max RTT surge and max RTT dip *in windows* for US servers and clients

```python
def chunk_list(lst, chunk_size):
    """Divide the list into chunks of specified size."""
    for i in range(0, len(lst), chunk_size):
        yield lst[i:i + chunk_size]

def find_minimums(lst, chunk_size):
    """Find minimums of chunks of the specified size in the list."""
    chunks = chunk_list(lst, chunk_size)
    min_values = [min(chunk) for chunk in chunks if len(chunk) == chunk_size]
    return min_values
```

```python
def compute_rtt_maxsurge_maxdip_in_windows(rtt_dict, win_size):
    maxsurge_perflow = {}
    maxdip_perflow = {}
    mismatch_count = 0
    for flow_id in rtt_dict:
        window_minrtts = find_minimums(rtt_dict[flow_id], win_size)
        if len(window_minrtts) < 2:
            continue
        first_rtt = window_minrtts[0]
        for second_rtt in window_minrtts[1:]:
            surge = second_rtt - first_rtt
            dip = first_rtt - second_rtt
            if surge > 0.0:
                if flow_id not in maxsurge_perflow:
                    maxsurge_perflow[flow_id] = surge
                elif surge > maxsurge_perflow[flow_id]:
                    maxsurge_perflow[flow_id] = surge
            if dip > 0.0:
                if flow_id not in maxdip_perflow:
                    maxdip_perflow[flow_id] = dip
                elif dip > maxdip_perflow[flow_id]:
                    maxdip_perflow[flow_id] = dip
            first_rtt = second_rtt + 0.0
        if flow_id in maxsurge_perflow and flow_id in maxdip_perflow:
            if maxsurge_perflow[flow_id] - maxdip_perflow[flow_id] > 1.0:
                mismatch_count += 1
    return maxsurge_perflow, maxdip_perflow, mismatch_count
```

```python
def compute_rtt_maxsurge_maxdip_metrics(win_size):
    win_maxsurge_perflow_pusrv_dict, win_maxdip_perflow_pusrv_dict, win_mismatch_count_pusrv = compute_rtt_maxsurge_maxdip_in_windows(
        rtts_usa_pusrv_dict, win_size)
    win_maxsurge_perflow_pucli_dict, win_maxdip_perflow_pucli_dict, win_mismatch_count_pucli = compute_rtt_maxsurge_maxdip_in_windows(
        rtts_usa_pucli_dict, win_size)
    win_maxsurge_perflow_pusrv = list(win_maxsurge_perflow_pusrv_dict.values())
    win_maxsurge_perflow_pucli = list(win_maxsurge_perflow_pucli_dict.values())
    win_maxsurge_perflow_all = win_maxsurge_perflow_pusrv[:] + win_maxsurge_perflow_pucli[:]
    win_maxdip_perflow_pusrv = list(win_maxdip_perflow_pusrv_dict.values())
    win_maxdip_perflow_pucli = list(win_maxdip_perflow_pucli_dict.values())
    win_maxdip_perflow_all = win_maxdip_perflow_pusrv[:] + win_maxdip_perflow_pucli[:]
    puall_maxsurge = [len(win_maxsurge_perflow_all),
                      np.percentile(win_maxsurge_perflow_all, 90.0),
                      np.percentile(win_maxsurge_perflow_all, 95.0),
                      np.percentile(win_maxsurge_perflow_all, 99.0),
                      np.percentile(win_maxsurge_perflow_all, 99.99)]
    pusrv_maxsurge = [len(win_maxsurge_perflow_pusrv),
                      np.percentile(win_maxsurge_perflow_pusrv, 90.0),
                      np.percentile(win_maxsurge_perflow_pusrv, 95.0),
                      np.percentile(win_maxsurge_perflow_pusrv, 99.0),
                      np.percentile(win_maxsurge_perflow_pusrv, 99.99)]
    pucli_maxsurge = [len(win_maxsurge_perflow_pucli),
                      np.percentile(win_maxsurge_perflow_pucli, 90.0),
                      np.percentile(win_maxsurge_perflow_pucli, 95.0),
                      np.percentile(win_maxsurge_perflow_pucli, 99.0),
                      np.percentile(win_maxsurge_perflow_pucli, 99.99)]
    puall_maxdip = [len(win_maxdip_perflow_all),
                      np.percentile(win_maxdip_perflow_all, 90.0),
                      np.percentile(win_maxdip_perflow_all, 95.0),
                      np.percentile(win_maxdip_perflow_all, 99.0),
                      np.percentile(win_maxdip_perflow_all, 99.99)]
    pusrv_maxdip = [len(win_maxdip_perflow_pusrv),
                      np.percentile(win_maxdip_perflow_pusrv, 90.0),
                      np.percentile(win_maxdip_perflow_pusrv, 95.0),
                      np.percentile(win_maxdip_perflow_pusrv, 99.0),
                      np.percentile(win_maxdip_perflow_pusrv, 99.99)]
    pucli_maxdip = [len(win_maxdip_perflow_pucli),
                      np.percentile(win_maxdip_perflow_pucli, 90.0),
                      np.percentile(win_maxdip_perflow_pucli, 95.0),
                      np.percentile(win_maxdip_perflow_pucli, 99.0),
                      np.percentile(win_maxdip_perflow_pucli, 99.99)]
    return puall_maxsurge, pusrv_maxsurge, pucli_maxsurge, puall_maxdip, pusrv_maxdip, pucli_maxdip
```

```python
# To vary: Window size
# To measure: Count of flows that qualify, p90, p95, p99, p99.99

puall_maxsurges = []
pusrv_maxsurges = []
pucli_maxsurges = []
puall_maxdips = []
pusrv_maxdips = []
pucli_maxdips = []

for window_size in [2,4,8,16,32,64,128,256,512]:
    print(f"Initiating for window size: {window_size}")
    puall_maxsurge_, pusrv_maxsurge_, pucli_maxsurge_, puall_maxdip_, pusrv_maxdip_, pucli_maxdip_ = compute_rtt_maxsurge_maxdip_metrics(window_size)
    puall_maxsurges.append(puall_maxsurge_)
    pusrv_maxsurges.append(pusrv_maxsurge_)
    pucli_maxsurges.append(pucli_maxsurge_)
    puall_maxdips.append(puall_maxdip_)
    pusrv_maxdips.append(pusrv_maxdip_)
    pucli_maxdips.append(pucli_maxdip_)
    print(f"Done for window size: {window_size}")
```
