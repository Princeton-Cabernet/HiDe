```python
countries = sorted(df_mindev_pred_mlab["Country1"].unique())

preattacks_mlab    = df_mindev_pred_mlab["PreAttack_p75_ms"].tolist()
postattacks_mlab   = df_mindev_pred_mlab["PostAttack_p25_ms"].tolist()

preattacks_campus  = df_mindev_pred_campus["PreAttack_p75_ms"].tolist()
postattacks_campus = df_mindev_pred_campus["PostAttack_p25_ms"].tolist()

x, y_mlab, y_campus = [], [], []

for p in range(1, 101, 1):
    preattack_mlab_at_p = np.percentile(preattacks_mlab, p)
    preattack_campus_at_p = np.percentile(preattacks_campus, p)
    
    defense_count_mlab = sum(1 for postattack in postattacks_mlab if postattack > preattack_mlab_at_p)
    defense_count_campus = sum(1 for postattack in postattacks_campus if postattack > preattack_campus_at_p)
    
    defense_pct_mlab = defense_count_mlab * 100.0 / len(postattacks_mlab)
    defense_pct_campus = defense_count_campus * 100.0 / len(postattacks_campus)
    
    x.append(p)
    y_mlab.append(defense_pct_mlab)
    y_campus.append(defense_pct_campus)

# # for i, j in zip(x, y):
# #     if round(j) >= 80 and round(j) <= 95:
# #         print(f"{round(j, 1)}% attacks > {i}%ile pre-attack RTT")

sns_colors = list(sns.color_palette("Reds_r"))

pu.lineplots([x, x], [y_campus, y_mlab], {
        "figsize": (8, 6),
        "colors": [sns_colors[0], sns_colors[3]],
        "ylim": (28, 102),
        "loc": "lower left",
        "curvelabels": ["Campus Data", "MLab Data"],
        "xlabel": "Pre-Attack MinRTT (%ile)",
        "ylabel": "Attack Scenarios (%)",
        "plot_path": "plots/comparison_cvg_prep75_postp25.pdf"
})
```

```python
mindevs_mlab   = df_mindev_pred_mlab["MinDev_p75_p25"].tolist()
mindevs_campus = df_mindev_pred_campus["MinDev_p75_p25"].tolist()

x, y_mlab, y_campus = [], [], []
for dev in range(0, 291, 1):
    defense_count_mlab   = sum(1 for mindev in mindevs_mlab if mindev > dev)
    defense_count_campus = sum(1 for mindev in mindevs_campus if mindev > dev)
    
    defense_pct_mlab   = defense_count_mlab * 100.0 / len(mindevs_mlab)
    defense_pct_campus = defense_count_campus * 100.0 / len(mindevs_campus)
    
    x.append(dev)
    y_mlab.append(defense_pct_mlab)
    y_campus.append(defense_pct_campus)

sns_colors = list(sns.color_palette("Blues_r"))

pu.lineplots([x, x], [y_campus, y_mlab], {
        "figsize": (8, 6),
        "colors": [sns_colors[0], sns_colors[3]],
        "ylim": (-5, 105),
        "xlim": (-10, 310),
        "loc": "upper right",
        "curvelabels": ["Campus Data", "MLab Data"],
        "xlabel": "Minimum $\delta_{deviation}$ (ms)",
        "ylabel": "Attack Scenarios (%)",
        "plot_path": "plots/comparison_mindev_prep75_postp25.pdf"
})
```

```python
countries = sorted(df_mindev_pred_campus['Country1'].unique())
coverages_campus, coverages_mlab = [], []
for country in countries:
    campus_pre  = df_mindev_pred_campus[df_mindev_pred_campus['Country1'] == country]['PreAttack_p75_ms'].tolist()
    campus_post = df_mindev_pred_campus[df_mindev_pred_campus['Country1'] == country]['PostAttack_p25_ms'].tolist()
    mlab_pre  = df_mindev_pred_mlab[df_mindev_pred_mlab['Country1'] == country]['PreAttack_p75_ms'].tolist()
    mlab_post = df_mindev_pred_mlab[df_mindev_pred_mlab['Country1'] == country]['PostAttack_p25_ms'].tolist()
    count_campus = sum(1 for pre, post in zip(campus_pre, campus_post) if post - pre >= 5)
    count_mlab = sum(1 for pre, post in zip(mlab_pre, mlab_post) if post - pre >= 5)
    coverages_campus.append(count_campus * 100.0 / len(campus_pre))
    coverages_mlab.append(count_mlab * 100.0 / len(mlab_pre))
coverages_campus = sorted(coverages_campus, reverse=True)
coverages_mlab   = sorted(coverages_mlab, reverse=True)

pct_countries      = []
pct_attacks_campus = []
pct_attacks_mlab   = []
for c in range(1, len(countries)+1):
    ac = coverages_campus[c-1]
    am = coverages_mlab[c-1]
    pct_countries.append(c * 100.0 / len(countries))
    pct_attacks_campus.append(ac)
    pct_attacks_mlab.append(am)

sns_colors = list(sns.color_palette("Blues_r"))
pu.lineplots([pct_countries]*2, [pct_attacks_campus, pct_attacks_mlab], {
        "figsize": (8, 6),
        "colors": [sns_colors[0], sns_colors[2]],
        "curvelabels": ["Campus data", "MLab data"],
        "loc": "lower left",
        "xlim": (-5, 105),
        "ylim": (-5, 105),
        "xlabel": "Countries (%)",
        "ylabel": "Optimal Attacks (%)",
        "plot_path": "plots/countries_vs_attacks_real.pdf"
})
```
